```python
price = float(input("Price:"))
quantity = int(input("Quantity: "))
has_coupon = input("Coupon? yes/no: ") == "yes"
is_member = input("Member? yes/no:") == "yes"

has_discount = has_coupon or is_member
print(has_discount)

subtotal = price * quantity
discount = subtotal * 0.10 if has_discount else 0
final_total = subtotal - discount

print(f"{final_total:.2f}")
```
